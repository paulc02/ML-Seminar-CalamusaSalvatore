### 🔹 Cerința 1: Funcții Python & Logică (2 puncte)
Scrieți o funcție Python numită `calculeaza_nota_finala(t1, t2, proiect)` care primește 3 note (numere reale între 1 și 10) și calculează media ponderată după următoarea regulă:
- **Test 1:** pondere de 30% (`0.30`)
- **Test 2:** pondere de 30% (`0.30`)
- **Proiect:** pondere de 40% (`0.40`)

*Cerință:* Funcția trebuie să returneze media ponderată rotunjită la două zecimale. Testați funcția pentru notele: `Test 1 = 8.5`, `Test 2 = 7.0`, `Proiect = 9.5`.

````python
# Rezolvare Cerința 1
def calculeaza_nota_finala(t1, t2, proiect):
    medie_ponderata = t1*0.3 + t2*0.3 + proiect*0.4
    return round(medie_ponderata, 2)

# Testare funcție
print(calculeaza_nota_finala(8.5, 7.0, 9.5))
````

### 🔹 Cerința 2: Operare cu Vectori în NumPy (3 puncte)
Avem următorul vector NumPy care reprezintă orele de studiu săptămânale ale 8 studenți:
```python
import numpy as np
ore_studiu = np.array([12, 5, 18, 8, 15, 3, 22, 10])
```

1. Calculați **media**, **deviația standard**, valoarea **minimă** și valoarea **maximă** a orelor de studiu folosind funcții din `numpy`.
2. Creați un filtru boolean care să identifice studenții care au studiat **mai mult de 10 ore** pe săptămână.
3. Afișați doar orele de studiu ale studenților care îndeplinesc această condiție.

````python
# Rezolvare Cerința 2
import numpy as np

ore_studiu = np.array([12, 5, 18, 8, 15, 3, 22, 10])

# 1. Statistici descriptive
media = np.mean(ore_studiu)
deviația_standard = np.std(ore_studiu)
minim = np.min(ore_studiu)
maxim = np.max(ore_studiu)

print(f"Media: {media}")
print(f"Deviația standard: {deviația_standard:.2f}")
print(f"Minim: {minim}")
print(f"Maxim: {maxim}")

# 2 & 3. Filtrare booleană
filtru_boolean = ore_studiu >= 10
ore_filtrate = ore_studiu[filtru_boolean]

print(f"Ore studiu filtrate: {ore_filtrate}")
````

### 🔹 Cerința 3: Creare și Manipulare Tabel Pandas (4 puncte)
Folosind dicționarul oferit, creați un `DataFrame` Pandas numit `df_studenti` și realizați următoarele prelucrări:

1. Adăugați o coloană nouă în tabel, numită `Nota_Finala`, calculată aplicând funcția `calculeaza_nota_finala` creată la **Cerința 1** peste coloanele `Test_1`, `Test_2` și `Proiect`.
2. Adăugați o coloană booleană `Admis` care conține valoarea `True` dacă `Nota_Finala >= 6.0` și `False` în caz contrar.
3. Filtrați tabelul și afișați doar studenții **Admiși** care au învățat **peste 10 ore** (`Ore_Studiu > 10`).

````python
# Rezolvare Cerința 3
import pandas as pd

date_studenti = {
    'Nume': ['Ana', 'Bogdan', 'Cristian', 'Elena', 'Florin', 'Gabriela', 'Horia', 'Ioana'],
    'Test_1': [8.5, 5.0, 9.0, 6.5, 4.0, 9.5, 7.0, 8.0],
    'Test_2': [7.0, 6.0, 8.5, 5.5, 4.5, 10.0, 6.5, 8.5],
    'Proiect': [9.5, 7.0, 9.5, 8.0, 5.0, 9.5, 7.5, 9.0],
    'Ore_Studiu': [12, 5, 18, 8, 15, 3, 22, 10]
}

df_studenti = pd.DataFrame(date_studenti)

# 1. Calcul Nota_Finala prin iterare rând cu rând
note_finale = []
for index, row in df_studenti.iterrows():
    nota = calculeaza_nota_finala(row['Test_1'], row['Test_2'], row['Proiect'])
    note_finale.append(nota)

df_studenti['Nota_Finala'] = note_finale

# 2. Adăugare coloană Admis
admis = []
for index, row in df_studenti.iterrows():
    if row['Nota_Finala'] >= 6.0:
        admis.append(True)
    else:
        admis.append(False)

df_studenti['Admis'] = admis

# 3. Filtrare
studenti_filtrati = df_studenti[(df_studenti['Admis'] == True) & (df_studenti['Ore_Studiu'] > 10)]

display(df_studenti)
display(studenti_filtrati)
````
